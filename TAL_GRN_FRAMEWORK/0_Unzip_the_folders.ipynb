```python
from google.colab import drive
drive.mount('/content/drive')

import zipfile, os

# ✅ One level deeper where the actual zips live
dataset_path = "/content/drive/MyDrive/Benchmark Diagnostic MRI and Medical Imaging Dataset/Medical Imaging Dataset"

for root, dirs, files in os.walk(dataset_path):
    for file in files:
        if file.endswith(".zip"):
            zip_path = os.path.join(root, file)
            folder_name = file.replace(".zip", "")
            extract_to = os.path.join(root, folder_name)

            if os.path.exists(extract_to):
                print(f"⏭ Skipping {folder_name}/ — already exists")
                continue

            os.makedirs(extract_to, exist_ok=True)
            with zipfile.ZipFile(zip_path, 'r') as z:
                z.extractall(extract_to)
            print(f"✓ {folder_name}/ extracted")

print("\nAll done!")
```

```python
print(f"{'Class':<80} {'Images':>8}")
print("-" * 90)

total = 0
for item in sorted(os.listdir(dataset_path)):
    item_path = os.path.join(dataset_path, item)
    if os.path.isdir(item_path):
        # Go one level deeper (nested same-name folder)
        nested_path = os.path.join(item_path, item)
        if os.path.exists(nested_path):
            count = len([f for f in os.listdir(nested_path)
                         if f.lower().endswith(('.jpg', '.jpeg', '.png'))])
        else:
            count = 0
        print(f"{item:<80} {count:>8}")
        total += count

print("-" * 90)
print(f"{'Total images':<80} {total:>8}")
```

```python
import os
import shutil

base_path = "/content/drive/MyDrive/Benchmark Diagnostic MRI and Medical Imaging Dataset/Medical Imaging Dataset"

for class_folder in os.listdir(base_path):

    class_path = os.path.join(base_path, class_folder)

    # skip if not folder
    if not os.path.isdir(class_path):
        continue

    inner_items = os.listdir(class_path)

    # check if only one inner folder exists
    if len(inner_items) == 1:

        inner_path = os.path.join(class_path, inner_items[0])

        if os.path.isdir(inner_path):

            # move files upward
            for file in os.listdir(inner_path):

                src = os.path.join(inner_path, file)
                dst = os.path.join(class_path, file)

                shutil.move(src, dst)

            # remove empty inner folder
            os.rmdir(inner_path)

            print("Fixed:", class_folder)
```
